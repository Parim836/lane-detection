```python
import os
import random
import shutil

# ===== CONFIG =====
DATASET_DIR = "/mnt/d/UNET++/lane"  # 🔴 แก้ path ให้ตรงของคุณ

TRAIN_IMG = os.path.join(DATASET_DIR, "train/images")
TRAIN_MASK = os.path.join(DATASET_DIR, "train/makes")

VAL_IMG = os.path.join(DATASET_DIR, "val/images")
VAL_MASK = os.path.join(DATASET_DIR, "val/makes")

MOVE_COUNT = 200
SEED = 42
# ==================

random.seed(SEED)

# สร้างโฟลเดอร์ val ถ้ายังไม่มี
os.makedirs(VAL_IMG, exist_ok=True)
os.makedirs(VAL_MASK, exist_ok=True)

# รายชื่อภาพทั้งหมดใน train/images
images = sorted(os.listdir(TRAIN_IMG))
random.shuffle(images)

moved = 0

for img_name in images:
    if moved >= MOVE_COUNT:
        break

    name, ext = os.path.splitext(img_name)

    img_src = os.path.join(TRAIN_IMG, img_name)
    mask_src = os.path.join(TRAIN_MASK, name + ".png")  # 🔴 แก้นามสกุล mask ถ้าไม่ใช่ png

    img_dst = os.path.join(VAL_IMG, img_name)
    mask_dst = os.path.join(VAL_MASK, name + ".png")

    if not os.path.exists(mask_src):
        print(f"⚠️ ไม่มี mask สำหรับ {img_name} → skip")
        continue

    shutil.move(img_src, img_dst)
    shutil.move(mask_src, mask_dst)

    moved += 1

print(f"✅ ย้ายไป val แล้ว {moved} คู่")
```

```python
class LaneDataset(Dataset):
    def __init__(self, images, masks):
        self.images = images
        self.masks = masks

    def __getitem__(self, idx):
        img = cv2.imread(self.images[idx])
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

        mask = cv2.imread(self.masks[idx], 0)

        # ⭐ สำคัญที่สุด
        img  = cv2.resize(img,  (512, 256))
        mask = cv2.resize(mask, (512, 256), interpolation=cv2.INTER_NEAREST)

        mask = (mask > 127).astype("float32")

        img = img.transpose(2,0,1) / 255.0

        return (
            torch.tensor(img, dtype=torch.float32),
            torch.tensor(mask, dtype=torch.float32).unsqueeze(0)
        )

    def __len__(self):
        return len(self.images)
```

```python
train_ds = LaneDataset(train_imgs, train_masks)
val_ds   = LaneDataset(val_imgs, val_masks)

train_loader = DataLoader(train_ds, batch_size=4, shuffle=True)
val_loader   = DataLoader(val_ds, batch_size=4)
```

```python
import torch

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Using device:", device)
```

```python
device = "cuda" if torch.cuda.is_available() else "cpu"

model = smp.DeepLabV3Plus(
    encoder_name="resnet34",
    encoder_weights="imagenet",
    encoder_output_stride=8,  
    in_channels=3,
    classes=1
).to(device) 

```

```python
import cv2
import torch
import matplotlib.pyplot as plt

# --- โหลดรูปใหม่ ---
test_img = cv2.imread("3.jpg")        # ← ใส่ path รูปของคุณตรงนี้
test_img = cv2.cvtColor(test_img, cv2.COLOR_BGR2RGB)
test_img = cv2.resize(test_img, (512,256))     # ขนาด input ของโมเดล

# --- แปลงเป็น tensor ---
x = torch.tensor(test_img.transpose(2,0,1)/255.).float().unsqueeze(0).to(device)

# --- Inference ก่อนเทรน ---
model.eval()
with torch.no_grad():
    pred = model(x)
    prob = pred.sigmoid()[0,0].cpu().numpy()

# --- Visualize ---
plt.figure(figsize=(10,4))
plt.subplot(1,2,1)
plt.imshow(test_img)
plt.title("Input Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(prob, cmap="gray")
plt.title("BEFORE TRAIN")
plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
dice_loss = smp.losses.DiceLoss(mode="binary")
bce_loss  = smp.losses.SoftBCEWithLogitsLoss()

def loss_fn(pred, target):
    return dice_loss(pred, target) + bce_loss(pred, target)
```

```python
val_loss = 0.0
iou_scores = []
dice_scores = []

with torch.no_grad():
    for imgs, masks in val_loader:
        imgs = imgs.float().to(device)
        masks = masks.float().to(device)

        preds = model(imgs)
        loss = loss_fn(preds, masks)
        val_loss += loss.item()

        probs = preds.sigmoid()
        pred_mask = (probs > 0.5).float()

        intersection = (pred_mask * masks).sum(dim=(1,2,3))
        union = (pred_mask + masks).sum(dim=(1,2,3)) - intersection

        iou = (intersection / (union + 1e-6)).mean().item()
        dice = (2 * intersection / ((pred_mask + masks).sum(dim=(1,2,3)) + 1e-6)).mean().item()

        iou_scores.append(iou)
        dice_scores.append(dice)

print(
    f"BEFORE TRAIN | "
    f"Loss: {val_loss/len(val_loader):.4f} | "
    f"IoU: {sum(iou_scores)/len(iou_scores):.4f} | "
    f"Dice: {sum(dice_scores)/len(dice_scores):.4f}"
)
```

```python
import torch

optimizer = torch.optim.AdamW(model.parameters(), lr=1e-4)
```

```python
def iou_score(pred, target, smooth=1e-6):
    pred = (pred > 0.5).float()
    intersection = (pred * target).sum(dim=(1,2,3))
    union = pred.sum(dim=(1,2,3)) + target.sum(dim=(1,2,3)) - intersection
    return ((intersection + smooth) / (union + smooth)).mean()

def dice_score(pred, target, smooth=1e-6):
    pred = (pred > 0.5).float()
    intersection = (pred * target).sum(dim=(1,2,3))
    return ((2*intersection + smooth) /
            (pred.sum(dim=(1,2,3)) + target.sum(dim=(1,2,3)) + smooth)).mean()
```

```python
def iou_score(pred, target, smooth=1e-6):
    # pred: sigmoid แล้ว (0–1)
    intersection = (pred * target).sum(dim=(1,2,3))
    union = pred.sum(dim=(1,2,3)) + target.sum(dim=(1,2,3)) - intersection
    return ((intersection + smooth) / (union + smooth)).mean()

def dice_score(pred, target, smooth=1e-6):
    intersection = (pred * target).sum(dim=(1,2,3))
    return ((2 * intersection + smooth) /
            (pred.sum(dim=(1,2,3)) + target.sum(dim=(1,2,3)) + smooth)).mean()
```

```python
best_iou = 0.0
num_epochs = 100

for epoch in range(num_epochs):

    # ===== TRAIN =====
    model.train()
    train_loss, train_iou, train_dice = 0, 0, 0

    for imgs, masks in train_loader:
        imgs, masks = imgs.to(device), masks.to(device)

        preds = model(imgs)
        loss = loss_fn(preds, masks)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        train_loss += loss.item()
        probs = torch.sigmoid(preds)
        train_iou += iou_score(probs, masks).item()
        train_dice += dice_score(probs, masks).item()

    train_loss /= len(train_loader)
    train_iou /= len(train_loader)
    train_dice /= len(train_loader)

    # ===== VALIDATION =====
    model.eval()
    val_loss, val_iou, val_dice = 0, 0, 0

    with torch.no_grad():
        for imgs, masks in val_loader:
            imgs, masks = imgs.to(device), masks.to(device)

            preds = model(imgs)
            loss = loss_fn(preds, masks)

            val_loss += loss.item()
            probs = torch.sigmoid(preds)
            val_iou += iou_score(probs, masks).item()
            val_dice += dice_score(probs, masks).item()

    val_loss /= len(val_loader)
    val_iou /= len(val_loader)
    val_dice /= len(val_loader)

    # ===== SAVE CHECKPOINT =====
    if val_iou > best_iou:
        best_iou = val_iou
        torch.save(model.state_dict(), "best_deeplab_lane.pth")
        print("✅ Save new best model")

    print(
        f"Epoch [{epoch+1}/{num_epochs}] | "
        f"Train IoU: {train_iou:.4f} | "
        f"Val IoU: {val_iou:.4f}"
    )
```

```python
def lane_accuracy(pred_bin, gt_bin, iou_thresh=0.3):
    pred_labels, pred_n = cluster_lanes(pred_bin)
    gt_labels, gt_n = cluster_lanes(gt_bin)

    if gt_n == 0 and pred_n == 0:
        return 1.0
    if gt_n == 0 and pred_n > 0:
        return 0.0

    matched_pred = set()
    tp = 0

    for gt_id in range(1, gt_n + 1):
        gt_mask = (gt_labels == gt_id)
        best_iou = 0.0
        best_pred_id = -1

        for pred_id in range(1, pred_n + 1):
            pred_mask = (pred_labels == pred_id)
            inter = np.logical_and(gt_mask, pred_mask).sum()
            union = np.logical_or(gt_mask, pred_mask).sum()
            iou = inter / (union + 1e-6)

            if iou > best_iou:
                best_iou = iou
                best_pred_id = pred_id

        if best_iou >= iou_thresh:
            tp += 1
            matched_pred.add(best_pred_id)

    fp = pred_n - len(matched_pred)
    fn = gt_n - tp

    acc = tp / (tp + fp + fn + 1e-6)
    return acc
```

```python
def lane_precision_recall_f1(pred_bin, gt_bin, iou_thresh=0.3):
    pred_labels, pred_n = cluster_lanes(pred_bin)
    gt_labels, gt_n = cluster_lanes(gt_bin)

    matched_pred = set()
    tp = 0

    for gt_id in range(1, gt_n + 1):
        gt_mask = (gt_labels == gt_id)

        best_iou = 0.0
        best_pred_id = -1

        for pred_id in range(1, pred_n + 1):
            pred_mask = (pred_labels == pred_id)
            inter = np.logical_and(gt_mask, pred_mask).sum()
            union = np.logical_or(gt_mask, pred_mask).sum()
            iou = inter / (union + 1e-6)

            if iou > best_iou:
                best_iou = iou
                best_pred_id = pred_id

        if best_iou >= iou_thresh:
            tp += 1
            matched_pred.add(best_pred_id)

    fn = gt_n - tp
    fp = pred_n - len(matched_pred)

    precision = tp / (tp + fp + 1e-6)
    recall = tp / (tp + fn + 1e-6)
    f1 = 2 * precision * recall / (precision + recall + 1e-6)

    return precision, recall, f1
```

```python
import numpy as np

def pixel_iou_and_dice(pred, gt, eps=1e-6):
    """
    pred, gt: numpy array (H, W) values {0,1}
    """
    pred = pred.astype(bool)
    gt = gt.astype(bool)

    intersection = np.logical_and(pred, gt).sum()
    union = np.logical_or(pred, gt).sum()

    iou = intersection / (union + eps)
    dice = (2.0 * intersection) / (pred.sum() + gt.sum() + eps)

    return iou, dice
```

```python
import cv2
import numpy as np

def cluster_lanes(binary, min_area=150):
    """
    binary: (H,W) uint8 {0,1}
    return:
        labels: (H,W) label image
        num_lanes: number of lanes
    """
    binary = (binary * 255).astype(np.uint8)

    num_labels, labels, stats, _ = cv2.connectedComponentsWithStats(
        binary, connectivity=8
    )

    filtered = np.zeros_like(binary)
    for i in range(1, num_labels):  # skip background
        if stats[i, cv2.CC_STAT_AREA] >= min_area:
            filtered[labels == i] = 255

    # optional morphology (เหมือนที่คุณใช้จริง)
    kernel = np.ones((4, 4), np.uint8)
    filtered = cv2.morphologyEx(filtered, cv2.MORPH_CLOSE, kernel)

    num_labels, labels = cv2.connectedComponents(filtered)
    num_lanes = num_labels - 1

    return labels, num_lanes
```

```python
model.eval()

iou_sum = dice_sum = 0.0
acc_sum = prec_sum = rec_sum = f1_sum = 0.0
count = 0

with torch.no_grad():
    for imgs, masks in val_loader:
        imgs = imgs.to(device)
        masks = masks.to(device)

        outputs = model(imgs)            # (B,1,H,W)
        probs = torch.sigmoid(outputs)

        B = probs.shape[0]

        for i in range(B):
            pred = probs[i, 0].cpu().numpy()   # (H,W)
            gt = masks[i, 0].cpu().numpy()     # (H,W)

            pred_bin = (pred > 0.5).astype(np.uint8)
            gt_bin = (gt > 0).astype(np.uint8)

            iou, dice = pixel_iou_and_dice(pred_bin, gt_bin)
            acc = lane_accuracy(pred_bin, gt_bin)
            precision, recall, f1 = lane_precision_recall_f1(pred_bin, gt_bin)

            iou_sum += iou
            dice_sum += dice
            acc_sum += acc
            prec_sum += precision
            rec_sum += recall
            f1_sum += f1
            count += 1

print("====== Evaluation (DeepLabV3) ======")
print(f"Pixel IoU        : {iou_sum / count:.4f}")
print(f"Pixel Dice       : {dice_sum / count:.4f}")
print(f"Lane Accuracy    : {acc_sum / count:.4f}")
print(f"Lane Precision   : {prec_sum / count:.4f}")
print(f"Lane Recall      : {rec_sum / count:.4f}")
print(f"Lane F1-score    : {f1_sum / count:.4f}")
print("===================================")
```
